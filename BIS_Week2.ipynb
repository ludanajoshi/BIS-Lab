```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import cross_val_score
from sklearn.neighbors import KNeighborsClassifier

# 1. Sample dataset (10 features: 5 useful, 5 noise)
Data, Labels = make_classification(n_samples=200, n_features=10, n_informative=5, random_state=42)

# 2. Initialize Swarm
M, N, MaxIter = 10, 10, 15
w, c1, c2 = 0.7, 1.5, 1.5
np.random.seed(42)

X = np.random.randint(0, 2, (N, M))
V = np.zeros((N, M))
pBest, pBest_score = X.copy(), np.full(N, -1.0)
gBest, gBest_score = X[0].copy(), -1.0

# 3. Main PSO Loop
for _ in range(MaxIter):
    for i in range(N):
        if np.sum(X[i]) == 0: X[i, 0] = 1
        acc = cross_val_score(KNeighborsClassifier(5), Data[:, X[i] == 1], Labels, cv=3).mean()

        if acc > pBest_score[i]: pBest_score[i], pBest[i] = acc, X[i].copy()
        if acc > gBest_score:    gBest_score, gBest = acc, X[i].copy()

    for i in range(N):
        V[i] = w*V[i] + c1*np.random.rand(M)*(pBest[i]-X[i]) + c2*np.random.rand(M)*(gBest-X[i])
        X[i] = (np.random.rand(M) < 1 / (1 + np.exp(-V[i]))).astype(int)


print(f"Best Mask: {gBest} | Selected: {np.sum(gBest)}/{M} | Accuracy: {gBest_score:.2%}")
```
